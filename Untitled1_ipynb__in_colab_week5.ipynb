````python
# Creating a stack
def create_stack():
    stack = []
    return stack

# Creating an empty stack
def check_empty(stack):
    return len(stack) == 0

# Adding items into the stack
def push(stack, item):
    stack.append(item)
    print("pushed item: " + item)

# Removing an element from the stack
def pop(stack):
    if (check_empty(stack)):
        return "stack is empty"
    return stack.pop()


# Function to reverse a string using a stack
def reverse_string(input_string):
    stack = create_stack()
    for char in input_string:
        push(stack, char)
    reversed_string = ""
    while not check_empty(stack):
        reversed_string += pop(stack)
    return reversed_string
#exaple usage
print("===REVERSING DATA USING STACK===")
#REVERS string
Text="hello world"
reversed_text=reverse_string(Text)
print(f"original string:{Text}")
print(f"reversed string:{reversed_text}")
print()
````

````python
# Creating a stack
def create_stack():
    stack = []
    return stack

# Creating an empty stack
def check_empty(stack):
    return len(stack) == 0

# Adding items into the stack
def push(stack, item):
    stack.append(item)
    print("pushed item: " + str(item)) # Convert item to string for printing

# Removing an element from the stack
def pop(stack):
    if (check_empty(stack)):
        return "stack is empty"
    return stack.pop()


# Function to reverse a list using a stack
def reverse_list(input_list):
    stack = create_stack()
    for item in input_list:
        push(stack, item)
    reversed_list = []
    while not check_empty(stack):
        reversed_list.append(pop(stack))
    return reversed_list
print("===REVERSING DATA USING STACK===")
#REVERS string
number=[1,2,3,4,5,6,7,8,9]
reversed_number = reverse_list(number)
print(f"original list:{number}")
print(f"reversed list:{reversed_number}")
print()
````

````python
import queue
import threading
import time

def demonstrate_builtin_queue():
  """ Demonstrate Python's built-in queue implementations """

  print("=== PYTHON BUILT-IN QUEUE ===")

  # 1. FIFO Queue
  print("\n1. FIFO Queue:")
  fifo_queue = queue.Queue(maxsize=3)

  # Add items
  for i in range(3):
    fifo_queue.put(i)
    print(f"Put: {i}")

  # Check if full
  print(f"Queue full: {fifo_queue.full()}")

  # Get items
  while not fifo_queue.empty():
    item = fifo_queue.get()
    print(f"Got: {item}")
    fifo_queue.task_done()

  # 2. LIFO Queue (Stack)
  print("\n2. LIFO Queue (Stack): ")
  lifo_queue = queue.LifoQueue()

  for i in range(3):
    lifo_queue.put(i)
    print(f"Pushed: {i}")

  while not lifo_queue.empty():
    item = lifo_queue.get()
    print(f"Popped: {item}")
````

````python
# First, we need a Node class to represent the 'node' struct in C
class Node:
    """Represents a single node in a binary tree."""
    def __init__(self, data):
        self.data = data   # The 'data' field
        self.left = None   # The 'left' pointer
        self.right = None  # The 'right' pointer

def inorder_traversal(root_node):
    """
    Performs a recursive inorder traversal (Left -> Root -> Right)
    of a binary tree and prints the data.
    """
    # Base case: if the node is None, do nothing.
    if root_node:
        # 1. Recursively visit the left subtree
        inorder_traversal(root_node.left)

        # 2. Visit the current node (print its data)
        # We use end=" " to print on the same line, just like the C code's space
        print(root_node.data, end=" ")

        # 3. Recursively visit the right subtree
        inorder_traversal(root_node.right)

# --- Example Usage ---
if __name__ == "__main__":

    # Let's build a simple tree manually:
    #      10
    #     /  \
    #    5    15
    #   /    /
    #  2    12

    root = Node(10)
    root.left = Node(5)
    root.right = Node(15)
    root.left.left = Node(2)
    root.right.left = Node(12)

    print("Starting Inorder Traversal (Left -> Root -> Right):")

    # Call the translated function
    inorder_traversal(root)

    print("\nTraversal complete.")

````

````python
from typing import List

# 3. Student Database
print("\n=== STUDENT DATABASE ===")

class Student:
    def __init__(self, student_id: str, name: str, grade: str, courses: List[str]):
        self.student_id = student_id
        self.name = name
        self.grade = grade
        self.courses = courses

    def __str__(self):
        return f"{self.name} (ID: {self.student_id}, Grade: {self.grade})"

# Create student database
student_db = dict()

# Add students
student_db["S001"] = Student("S001", "Alice Johnson", "A", ["Math", "Physics", "CS"])
student_db["S002"] = Student("S002", "Bob Smith", "B", ["History", "English"])
student_db["S003"] = Student("S003", "Carol Davis", "A", ["Biology", "Chemistry"])

# Query students
print("All Students:")
for student_id, student in student_db.items():
    print(f"  {student_id}: {student}")

# Find specific student
if "S002" in student_db:
    student = student_db["S002"]
    print(f"\nFound student: {student.name}")
    print(f"Courses: {', '.join(student.courses)}")

# 4. Product Inventory System
print("\n=== PRODUCT INVENTORY ===")

class Product:
    def __init__(self, product_id: str, name: str, price: float, quantity: int):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.quantity = quantity

    def __str__(self):
        return f"{self.name} - ${self.price} (Qty: {self.quantity})"

inventory = dict()

# Add products
inventory["P100"] = Product("P100", "Laptop", 999.99, 10)
inventory["P101"] = Product("P101", "Mouse", 25.50, 50)
inventory["P102"] = Product("P102", "Keyboard", 75.00, 25)
inventory["P103"] = Product("P103", "Monitor", 299.99, 15)

# Inventory operations
def display_inventory(inv: dict) -> None:
    print("Current Inventory:")
    for product_id, product in inv.items():
        print(f"  {product_id}: {product}")

def update_stock(inv: dict, product_id: str, quantity_change: int) -> None:
    if product_id in inv:
        product = inv[product_id]
        product.quantity += quantity_change
        if product.quantity < 0:
            product.quantity = 0
        print(f"Updated {product.name}: New quantity = {product.quantity}")
    else:
        print(f"Product {product_id} not found")

display_inventory(inventory)

# Simulate sales and restocking
print("\n=== INVENTORY UPDATES ===")
update_stock(inventory, "P100", -2)  # Sell 2 laptops
update_stock(inventory, "P101", 10)  # Restock 10 mice
update_stock(inventory, "P102", -5)  # Sell 5 keyboards

display_inventory(inventory)

# 5. Configuration Management
print("\n=== CONFIGURATION MANAGEMENT ===")

app_config = dict()

# Application settings
app_config["database"] = {
    "host": "localhost",
    "port": 5432,
    "name": "myapp_db",
    "user": "admin"
}

app_config["server"] = {
    "host": "0.0.0.0",
    "port": 8080,
    "debug": True
}

app_config["features"] = {
    "authentication": True,
    "caching": False,
    "logging": True
}

# Access configuration
db_config = app_config["database"]
print(f"Database: {db_config['host']}:{db_config['port']}")
print(f"Server port: {app_config['server']['port']}")
print(f"Authentication enabled: {app_config['features']['authentication']}")

# Update configuration
app_config["server"]["debug"] = False
print(f"Debug mode: {app_config['server']['debug']}")
````
